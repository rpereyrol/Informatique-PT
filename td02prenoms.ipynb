## Exercice 5

#### 1. Écrire une requête SQL qui renvoie la liste des prénoms les plus donnés en 2020.

```sql
SELECT prenom, SUM(nombre) AS naissances
FROM prenoms
WHERE annais=2020
GROUP BY prenom
ORDER BY naissances DESC
LIMIT 20
```

#### 2.  Écrire une requête SQL qui renvoie le nombre de naissances en 2018.

```sql
SELECT SUM(nombre)
FROM prenoms
WHERE annais=2018
```

#### 3. Écrire une requête SQL qui renvoie les trois années où le prénom DOMINIQUE a été le plus donné.

!!! warning

Attention aux prénoms épicènes, attribués à des garçons et des filles comme par exemple Dominique.
Il faut compter le nombre de files **et** de garçons pour ces prénoms.
!!!

```sql
SELECT annais FROM prenoms
WHERE prenom='DOMINIQUE'
GROUP BY annais
ORDER BY SUM(nombre) DESC
LIMIT 3
```

#### 4. Écrire une requête SQL qui renvoie les prénoms attribués aux deux sexes.

```sql
SELECT prenom
FROM prenoms
GROUP BY prenom
HAVING count(DISTINCT sexe)>1
```

## Exercice 6

#### 1. Écrire une requête SQL qui renvoie, pour chaque élève de la classe, le nombre d'enfants nés la même année que lui.

```sql
SELECT c.prenom, c.annais, SUM(p.nombre)
FROM classe c
JOIN prenoms p
ON c.annais=p.annais
GROUP BY c.id
```

#### 2. Écrire une requête SQL qui renvoie, pour chaque élève de la classe, le nombre d'enfants ayant le même prénom nés la même année que lui.

```sql
SELECT c.id, c.prenom, c.annais, SUM(p.nombre)
FROM classe c
JOIN prenoms p
ON c.annais=p.annais
    AND c.prenom=p.prenom
GROUP BY c.id
```

#### 3. Écrire une requête SQL qui renvoie, pour chaque prénom de la classe la première année où ce prénom est apparu (dans la base de données).

```sql
SELECT DISTINCT c.prenom, MIN(p.annais)
FROM classe c
JOIN prenoms p
ON c.prenom=p.prenom
GROUP BY c.prenom
```
